```python
df.isnull().sum()
```

I checked for missing values using df.isnull().sum().
Only Education and AnnualIncome columns have missing values, with 12 entries missing each.
All other columns are complete, so no cleaning is needed for them.

```python
# Fill missing values
df['Education'].fillna(df['Education'].mode()[0], inplace=True)
df['AnnualIncome'].fillna(df['AnnualIncome'].median(), inplace=True)
```

For Education (categorical), I filled missing values with the mode because it preserves the most common category and maintains the original data distribution.
For AnnualIncome (numerical), I filled missing values with the median to avoid the influence of extreme high-income values (outliers).

```python
df.duplicated().sum()
```

I checked for duplicate rows using df.duplicated().sum() and found 5 duplicates.
Duplicate rows can skew analysis results, so it is important to remove them to ensure each customer is represented uniquely.

```python
df.drop_duplicates(inplace=True)
```

Duplicate rows were removed using df.drop_duplicates().
The dataset now contains only unique customer entries.

```python
df.isnull().sum()
df.duplicated().sum()
```

After preprocessing, all missing values were filled and all duplicate rows were removed.
The dataset is now clean and ready for analysis.

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=10, kde=True, color='skyblue')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

### Insight:
Most customers are aged between 29 and 46, indicating that the majority of the customer base consists of working adults. There are fewer customers younger than 29 or older than 46.

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Gender', data=df, palette='pastel')
plt.title('Gender Distribution of Customers')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

### Insight:
The customer base has slightly more male customers (128) than female customers (122). Although the difference is small, gender may still influence purchasing patterns and could be explored further.

```python
plt.figure(figsize=(8,5))
sns.histplot(df['SpendingScore'], bins=10, kde=True, color='salmon')
plt.title('Distribution of Spending Score')
plt.xlabel('Spending Score')
plt.ylabel('Count')
plt.show()
```

### Insight:
Spending scores range from 5 to 95, with the majority of customers scoring between the 25th percentile (35) and 75th percentile (58). This indicates a mix of low, medium, and high spenders, reflecting varied purchasing behavior.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='AnnualIncome', y='SpendingScore', hue='Gender', data=df, palette='Set2')
plt.title('Annual Income vs Spending Score by Gender')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.show()
```

### Insight:
Spending scores are spread across all income levels, showing no strong correlation between income and spending behavior. Gender does not show a clear pattern either, meaning both males and females have varied spending scores regardless of income.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Age', y='PurchaseFrequency', data=df)
plt.title('Age vs Purchase Frequency')
plt.xlabel('Age')
plt.ylabel('Purchase Frequency')
plt.show()
```

### Insight:
Purchase frequency varies across ages, with notable peaks at ages 21, 35, 43, 49, and 50. Some mid-age customers (25–45) show moderate to low purchasing activity, suggesting that age alone may not fully explain purchase patterns. Further analysis, possibly grouping ages into bins, could reveal clearer trends.

```python
# Compute correlation matrix only for numeric columns
corr = df.corr(numeric_only=True)

# Plot heatmap
plt.figure(figsize=(10,7))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

### Insight:
Age and YearsEmployed are strongly positively correlated, as expected. Older customers tend to have more work experience.

AnnualIncome and LastPurchaseAmount are essentially uncorrelated, indicating that higher income does not necessarily translate to larger purchases.

PurchaseFrequency and OnlineVisitsPerMonth have almost no correlation, suggesting that visiting the site more frequently doesn’t directly lead to more purchases.

SpendingScore shows weak to moderate correlations with other variables, implying that spending behavior depends on multiple factors beyond just income or age.
