## Machine Learning Categories

### Supervised Learning

Supervised learning is a type of machine learning where the dataset contains input data along with correct output labels. The model learns from these examples to predict results for new data.

### Unsupervised Learning

Unsupervised learning works with data that has no labels. The algorithm automatically finds hidden patterns or groups in the data.

### Reinforcement Learning

Reinforcement learning is a learning method where an agent learns by interacting with an environment and improves its performance using rewards and penalties.
